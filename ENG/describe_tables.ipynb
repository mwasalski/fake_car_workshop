# DESCRIBE EXTENDED – fake_car_workshop

Runs `DESCRIBE EXTENDED` for every table in the catalog.  
Set the catalog name in the widget below before running.

```python
# Widget – set catalog name before running
dbutils.widgets.text('catalog', 'fake_car_workshop_franchise', 'Catalog')
CATALOG = dbutils.widgets.get('catalog')

print(f'Catalog: {CATALOG}')
print(f'dim  schema: {CATALOG}.dim')
print(f'fact schema: {CATALOG}.fact')
```
